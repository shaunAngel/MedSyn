```python
#Python program to check whether an element exists within a tuple
tup=(1,2,3,4,5)
x=2
if x in tup:
    print("Element in tuple")
else:
    print("Element not in tuple")
```

```python
#Python program to find maximum and minimum element in tuple
tup=(1,3,5,7,9)
print("Maximum=",max(tup))
print("Minimum=",min(tup))
```

```python
#Python program to sort a tuple
tup=(3,5,9,4,7)
tup1=tuple(sorted(tup))
print("Sorted tuple=",tup1)
```

```python
#Python program to remove duplicates from tuple
tup1=(1,2,2,3,4,5,5)
tup2=tuple(set(tup1))
print("Tuple with no duplicates=",tup2)
```

```python
#Python program to create a nested tuple and access elements
tup=((1,2),(3,4),(5,6))
print(tup[0])
print(tup[2])
print(tup[1][0])
```

```python
#Python program to unpack tuple into variables
tup=(10,20,30)
a,b,c=tup
print(a,b,c)
```

```python
#Python program to count frequency of elements in tuple
t=(1,2,2,3,3,3,4)
print("Frequency of 3=", t.count(3))
```

```python
#Python program to find repeated elements in tuple
t=(1,2,2,3,3,3,4,4)
notrep=[]
rep=[]
for x in t:
    if x in notrep:
        if x not in rep:
            rep.append(x)
    else:
        notrep.append(x)

print("Repeated elements=",rep)
```

```python
#Python program to convert a list and tuple into arrays
from array import array
L=[1,2,3,4]
tup=(5,6,7,8)
arr1=array('i',L)
arr2=array('i',tup)
print("Array from list:",arr1)
print("Array from tuple:",arr2)
```

```python
#Python program to find common values between two arrays
a=[1, 2, 3, 4, 5]
b=[3, 4, 5, 6, 7]
print("Common values are:")
for i in a:
    if i in b:
        print(i)
```
